## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

```python
# ML-10 — Ranked actions + reason codes

import os
import pandas as pd

REPO = "/content/flyrank-ml-internship-starter"

if not os.path.exists(REPO):
    !git clone -q https://github.com/Sameer99-star/flyrank-ml-internship-starter.git

DATA_PATH = f"{REPO}/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(DATA_PATH)

# Recreate the validated baseline queue from ML-07.
ranked = df.copy()

ranked["score"] = (
    ranked["days_since_last_update"].fillna(0).clip(lower=0) / 30
    + ranked["trend_pct"].fillna(0).clip(upper=0).abs() / 10
)

ranked["reason_code"] = "AGE_AND_DECLINE"
ranked["action"] = "REFRESH"

ranked = ranked.sort_values(
    ["score", "days_since_last_update"],
    ascending=[False, False]
).reset_index(drop=True)

ranked["rank"] = ranked.index + 1

# Human-readable action queue.
action_queue = ranked[
    [
        "rank",
        "content_id",
        "score",
        "action",
        "reason_code",
        "freshness_tier",
        "days_since_last_update",
        "trend_direction",
        "trend_pct",
        "ctr",
        "avg_position",
    ]
].copy()

print("Ranked action queue created from the ML-07 baseline.")
print(f"Rows: {len(action_queue)}")
print("\nTop 10 actions:")
display(action_queue.head(10))
```

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

```python
# ML-10 — Intended use and limits

print("""
INTENDED USE
------------
This action queue is intended to help content and SEO teams prioritize
pages for human review and possible refresh.

The queue ranks content using the validated ML-07 baseline signals:
- days_since_last_update
- trend_pct

The output is a prioritization aid, not an automatic publishing system.

LIMITS
------
- The score is a simple baseline, not a causal prediction.
- A high score does not prove that refreshing a page will improve performance.
- The queue does not replace editorial, SEO, or business judgment.
- The available signals may not capture important context such as seasonality,
  search intent changes, technical SEO issues, or planned content changes.
- All recommended actions require human review before execution.
- The model should not be used to automatically rewrite, publish, delete,
  or redirect content.

VALIDITY BOUNDARY
-----------------
This playbook is most useful for prioritizing pages with observable
performance decline and/or content age. It should be treated as a
decision-support tool and revalidated if the data distribution,
business objective, or content strategy changes.
""")
```

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

```python
# ML-10 — Human review + the no-go list

print("""
HUMAN REVIEW RULES
------------------
Every ranked recommendation must be reviewed by a human before action.

For each candidate, the reviewer should check:
1. Whether the content is still relevant to the target audience.
2. Whether search intent has changed.
3. Whether the decline is meaningful rather than temporary.
4. Whether there are technical SEO or measurement issues.
5. Whether the page has business, legal, or editorial constraints.
6. Whether refreshing the page is more valuable than other available work.

NO-GO LIST
----------
The system should NOT automatically:
- Rewrite or publish content.
- Delete content.
- Redirect URLs.
- Change important business or editorial claims.
- Make legal, medical, financial, or compliance decisions.
- Treat a high score as proof that a refresh will improve performance.

FINAL DECISION
--------------
The model ranks candidates for review. A qualified human makes the
final decision about whether to refresh, monitor, investigate, or take
no action.
""")
```

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

```python
# ML-10 — Monitoring / retrain triggers

print("""
MONITORING
----------
The action queue should be checked periodically for changes in data
quality and recommendation behavior.

Monitor:
- Distribution of days_since_last_update.
- Distribution of trend_pct.
- Number and proportion of pages receiving REFRESH recommendations.
- Missing values in the signals used by the baseline.
- Changes in the content mix or traffic environment.

RETRAIN / REVALIDATION TRIGGERS
-------------------------------
Revalidate the baseline when:
1. The input data distribution changes materially.
2. Missingness in key signals increases.
3. The proportion of recommended refreshes changes unexpectedly.
4. Content strategy, search behavior, or business objectives change.
5. Human reviewers repeatedly reject the recommended action.

A retrain or model update should only happen after reviewing the cause
of the change and confirming that the underlying data and objective
are still appropriate.
""")
```
